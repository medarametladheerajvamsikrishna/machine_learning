Question: For the below dataset, identify which variable is principal component.
| Variable       |     1 |      2 |      3 |
| -------------- | ----: | -----: | -----: |
| Climate        | 0.190 |  0.017 |  0.207 |
| Housing        | 0.544 |  0.020 |  0.204 |
| Health         | 0.782 | -0.605 |  0.144 |
| Crime          | 0.365 |  0.294 |  0.585 |
| Transportation | 0.585 |  0.085 |  0.234 |
| Education      | 0.394 | -0.273 |  0.027 |
| Arts           | 0.985 |  0.126 | -0.111 |
| Recreation     | 0.520 |  0.402 |  0.519 |
| Economy        | 0.142 |  0.150 |  0.239 |


```python
import pandas as pd
import numpy as np
```

```python
data = {
    'Variable': [
        'Climate',
        'Housing',
        'Health',
        'Crime',
        'Transportation',
        'Education',
        'Arts',
        'Recreation',
        'Economy'
    ],

    'PC1': [
        0.190,
        0.544,
        0.782,
        0.365,
        0.585,
        0.394,
        0.985,
        0.520,
        0.142
    ],

    'PC2': [
        0.017,
        0.020,
        -0.605,
        0.294,
        0.085,
        -0.273,
        0.126,
        0.402,
        0.150
    ],

    'PC3': [
        0.207,
        0.204,
        0.144,
        0.585,
        0.234,
        0.027,
        -0.111,
        0.519,
        0.239
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
pc1_index = df['PC1'].abs().idxmax()

print("PC1 variable:", df.loc[pc1_index, 'Variable'])
print("PC1 loading:", df.loc[pc1_index, 'PC1'])
```

```python
pc2_index = df['PC2'].abs().idxmax()

print("PC2 variable:", df.loc[pc2_index, 'Variable'])
print("PC2 loading:", df.loc[pc2_index, 'PC2'])
```

```python
pc3_index = df['PC3'].abs().idxmax()

print("PC3 variable:", df.loc[pc3_index, 'Variable'])
print("PC3 loading:", df.loc[pc3_index, 'PC3'])
```

```python
for pc in ['PC1', 'PC2', 'PC3']:
    index = df[pc].abs().idxmax()

    print(
        pc,
        "->",
        df.loc[index, 'Variable'],
        "-> Loading:",
        df.loc[index, pc]
    )
```

```python
result = []

for pc in ['PC1', 'PC2', 'PC3']:
    index = df[pc].abs().idxmax()

    result.append([
        pc,
        df.loc[index, 'Variable'],
        df.loc[index, pc]
    ])

result_df = pd.DataFrame(
    result,
    columns=['Principal Component', 'Variable', 'Loading']
)

print(result_df)
```

```python
import matplotlib.pyplot as plt

variables = ['Arts', 'Health', 'Crime']
loadings = [0.985, -0.605, 0.585]

plt.bar(['PC1', 'PC2', 'PC3'], loadings)

plt.xlabel('Principal Components')
plt.ylabel('Loading')
plt.title('Highest Loading Variable for Each Principal Component')

plt.show()
```

Thus, the principal components were identified based on the highest absolute loading values. PC1 corresponds to Arts (0.985), PC2 corresponds to Health (-0.605), and PC3 corresponds to Crime (0.585).
